```python

import cv2
import numpy as np
import matplotlib.pyplot as plt

img = np.ones((400, 600, 3), dtype=np.uint8) * 255
cv2.rectangle(img, (50, 50), (250, 250), (0, 0, 255), -1)
cv2.circle(img, (400, 150), 80, (255, 0, 0), -1)
cv2.rectangle(img, (300, 270), (550, 350), (0, 255, 0), -1)
cv2.putText(img, "OPENCV", (150, 390), cv2.FONT_HERSHEY_SIMPLEX, 1, (0, 0, 0), 2)

def show(title, image):
    if len(image.shape) == 2:
        plt.imshow(image, cmap="gray")
    else:
        plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
    plt.title(title)
    plt.axis("off")
    plt.show()

show("Sample Original Image", img)
```

```python

gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
show("Original Image", img)
show("Grayscale Image", gray)
```

```python

blur = cv2.GaussianBlur(img, (15, 15), 0)
show("Original Image", img)
show("Gaussian Blur Image", blur)
```

```python

gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
edges = cv2.Canny(gray, 100, 200)
show("Original Image", img)
show("Canny Edge Detection", edges)
```

```python

gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
equalized = cv2.equalizeHist(gray)
show("Original Grayscale Image", gray)
show("Histogram Equalized Image", equalized)
```

```python

colors = ("b", "g", "r")
for i, color in enumerate(colors):
    hist = cv2.calcHist([img], [i], None, [256], [0, 256])
    plt.plot(hist, color=color)
plt.title("Color Histogram")
plt.xlabel("Pixel Intensity")
plt.ylabel("Number of Pixels")
plt.xlim([0, 256])
plt.show()
```

```python

kernel = np.ones((5, 5), np.uint8)
eroded = cv2.erode(img, kernel, iterations=1)
show("Original Image", img)
show("Eroded Image", eroded)
```

```python

frames = []
for i in range(30):
    frame = np.zeros((300, 500, 3), dtype=np.uint8)
    cv2.circle(frame, (30 + i * 14, 150), 35, (0, 255, 255), -1)
    cv2.putText(frame, "Sample Video", (130, 50), cv2.FONT_HERSHEY_SIMPLEX, 1, (255, 255, 255), 2)
    frames.append(frame)

def play_video(frames, delay, title):
    for frame in frames:
        plt.figure(figsize=(6, 3))
        plt.imshow(cv2.cvtColor(frame, cv2.COLOR_BGR2RGB))
        plt.title(title)
        plt.axis("off")
        plt.show()
        plt.close()

play_video(frames, 200, "Slow Motion Video")
play_video(frames, 1, "Fast Motion Video")
```

```python

kernel = np.ones((5, 5), np.uint8)
dilated = cv2.dilate(img, kernel, iterations=1)
show("Original Image", img)
show("Dilated Image", dilated)
```

```python

height, width = img.shape[:2]
bigger = cv2.resize(img, (width * 2, height * 2))
smaller = cv2.resize(img, (width // 2, height // 2))
show("Original Image", img)
show("Bigger Image", bigger)
show("Smaller Image", smaller)
```

```python

rotated = cv2.rotate(img, cv2.ROTATE_90_CLOCKWISE)
show("Original Image", img)
show("90 Degree Clockwise Rotation", rotated)
```
