```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = np.zeros((300, 400, 3), dtype=np.uint8)
img[:] = (100, 100, 100)
cv2.rectangle(img, (20, 30), (130, 270), (255, 0, 0), -1)
cv2.rectangle(img, (140, 30), (260, 270), (0, 255, 0), -1)
cv2.rectangle(img, (270, 30), (390, 270), (0, 0, 255), -1)

colors = ('b', 'g', 'r')
labels = ('Blue', 'Green', 'Red')

for i, color in enumerate(colors):
    hist = cv2.calcHist([img], [i], None, [256], [0, 256])
    plt.plot(hist, color=color, label=labels[i])

plt.title("Color Histogram")
plt.xlabel("Pixel Intensity")
plt.ylabel("Number of Pixels")
plt.legend()
plt.xlim([0, 256])
plt.show()
```

```python
import cv2
import numpy as np
from IPython.display import display, clear_output
from PIL import Image
import time

frames = []
for i in range(20):
    frame = np.zeros((250, 400, 3), dtype=np.uint8)
    frame[:] = (255, 255, 255)
    cv2.circle(frame, (30 + i*15, 125), 25, (0, 0, 255), -1)
    frames.append(frame)

def play_video(speed, title):
    for frame in frames:
        clear_output(wait=True)
        print(title)
        display(Image.fromarray(cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)))
        time.sleep(speed)

play_video(0.15, "Slow Motion Video")
play_video(0.02, "Fast Motion Video")
```

```python
import cv2
import numpy as np
from google.colab.patches import cv2_imshow

img = np.zeros((300, 400, 3), dtype=np.uint8)
img[:] = (255, 255, 255)
cv2.rectangle(img, (50, 50), (200, 250), (0, 0, 255), -1)

rows, cols = img.shape[:2]
src = np.float32([[0, 0], [cols-1, 0], [0, rows-1]])
dst = np.float32([[50, 50], [cols-50, 30], [80, rows-50]])

matrix = cv2.getAffineTransform(src, dst)
result = cv2.warpAffine(img, matrix, (cols, rows))

print("Original Image:")
cv2_imshow(img)
print("Affine Transformation:")
cv2_imshow(result)
```

```python
import cv2
import numpy as np
from google.colab.patches import cv2_imshow

img = np.zeros((300, 400, 3), dtype=np.uint8)
img[:] = (255, 255, 255)
cv2.rectangle(img, (50, 50), (200, 250), (0, 0, 0), 3)
cv2.rectangle(img, (250, 80), (350, 220), (0, 0, 0), 3)

gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
gray = np.float32(gray)

corners = cv2.cornerHarris(gray, 2, 3, 0.04)
corners = cv2.dilate(corners, None)

result = img.copy()
result[corners > 0.01 * corners.max()] = [0, 0, 255]

print("Harris Corner Detection:")
cv2_imshow(result)
```

```python
import cv2
import numpy as np
from google.colab.patches import cv2_imshow

img = np.zeros((300, 400, 3), dtype=np.uint8)
img[:] = (255, 255, 255)
cv2.rectangle(img, (50, 50), (200, 250), (0, 0, 0), -1)
cv2.circle(img, (280, 150), 70, (0, 0, 0), -1)

gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

sobel_x = cv2.Sobel(gray, cv2.CV_64F, 1, 0, ksize=3)
sobel_y = cv2.Sobel(gray, cv2.CV_64F, 0, 1, ksize=3)

sobel_x = cv2.convertScaleAbs(sobel_x)
sobel_y = cv2.convertScaleAbs(sobel_y)

result = cv2.addWeighted(sobel_x, 0.5, sobel_y, 0.5, 0)

print("Original Image:")
cv2_imshow(img)
print("Sobel X:")
cv2_imshow(sobel_x)
print("Sobel Y:")
cv2_imshow(sobel_y)
print("Sobel Result:")
cv2_imshow(result)
```
