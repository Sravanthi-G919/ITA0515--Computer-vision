```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = np.zeros((300, 400, 3), dtype=np.uint8)
img[:] = (100, 100, 100)
cv2.rectangle(img, (20, 30), (130, 270), (255, 0, 0), -1)
cv2.rectangle(img, (140, 30), (260, 270), (0, 255, 0), -1)
cv2.rectangle(img, (270, 30), (390, 270), (0, 0, 255), -1)

colors = ('b', 'g', 'r')
labels = ('Blue', 'Green', 'Red')

for i, color in enumerate(colors):
    hist = cv2.calcHist([img], [i], None, [256], [0, 256])
    plt.plot(hist, color=color, label=labels[i])

plt.title("Color Histogram")
plt.xlabel("Pixel Intensity")
plt.ylabel("Number of Pixels")
plt.legend()
plt.xlim([0, 256])
plt.show()
```

```python
import cv2
import numpy as np
from google.colab.patches import cv2_imshow

img = np.zeros((300, 400), dtype=np.uint8)
cv2.rectangle(img, (80, 60), (300, 240), 255, -1)
cv2.circle(img, (200, 150), 60, 0, -1)

kernel = np.ones((5, 5), np.uint8)
eroded = cv2.erode(img, kernel, iterations=1)

print("Original Image:")
cv2_imshow(img)
print("Eroded Image:")
cv2_imshow(eroded)
```

```python
import cv2
import numpy as np
from IPython.display import display, clear_output
from PIL import Image
import time

frames = []
for i in range(20):
    frame = np.zeros((250, 400, 3), dtype=np.uint8)
    frame[:] = (255, 255, 255)
    cv2.circle(frame, (30 + i*15, 125), 25, (0, 0, 255), -1)
    frames.append(frame)

def play_video(speed, title):
    for frame in frames:
        clear_output(wait=True)
        print(title)
        display(Image.fromarray(cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)))
        time.sleep(speed)

play_video(0.15, "Slow Motion Video")
play_video(0.02, "Fast Motion Video")
```

```python
import cv2
import numpy as np
from google.colab.patches import cv2_imshow

img = np.zeros((300, 400), dtype=np.uint8)
cv2.rectangle(img, (100, 80), (250, 220), 255, -1)

kernel = np.ones((5, 5), np.uint8)
dilated = cv2.dilate(img, kernel, iterations=2)

print("Original Image:")
cv2_imshow(img)
print("Dilated Image:")
cv2_imshow(dilated)
```

```python
import cv2
import numpy as np
from google.colab.patches import cv2_imshow

img = np.zeros((200, 300, 3), dtype=np.uint8)
img[:] = (0, 255, 255)
cv2.circle(img, (150, 100), 60, (0, 0, 255), -1)

bigger = cv2.resize(img, (600, 400))
smaller = cv2.resize(img, (150, 100))

print("Original Image:")
cv2_imshow(img)
print("Bigger Image:")
cv2_imshow(bigger)
print("Smaller Image:")
cv2_imshow(smaller)
```
